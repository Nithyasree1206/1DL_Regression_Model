```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
X=torch.linspace(1,50,50).reshape(-1,1)
```

```python
torch.manual_seed(71)
e=torch.randint(-8,9,(50,1),dtype=torch.float)
```

```python
y=2*X+1+e
```

```python
plt.scatter(X,y)
plt.ylable("y")
plt.xlabel("x");
```

```python
class Model(nn.Module):
    def __init__(self, in_features, out_features):
        super().__init__()
        self.linear = nn.Linear(in_features, out_features)
        
    def forward(self, x):
        y_pred = self.linear(x)
        return y_pred
```

```python
torch.manual_seed(59)
model = Model(1, 1)
```

```python
model.linear.weight.item()
```

```python
model.linear.bias.item()
```

```python
x=torch.tensor([50])
```

```python
x
```

```python
x=x.float()
model.forward(x)
```

```python
x1=np.linspace(0.0,50.0,50)
```

```python
x1
```

```python
w1=0.1059
b1=0.9637
```

```python
y1=w1*x1+b1
```

```python
y1
```

```python
plt.plot(x1,y1,'r')
plt.scatter(X,y)
```

```python
loss_function = nn.MSELoss()  # Mean Squared Error (MSE) loss

optimizer = torch.optim.SGD(model.parameters(), lr=0.0001)  # Stochastic Gradient Descent

epochs = 50  # Number of training iterations
losses = []  # List to store loss values

for epoch in range(1, epochs + 1):  # Start from 1 to 50
    optimizer.zero_grad()  # Clear previous gradients
    y_pred = model(X)  # Forward pass
    loss = loss_function(y_pred, y)  # Compute loss
    losses.append(loss.item())  # Store loss value
    
    loss.backward()  # Compute gradients
    optimizer.step()  # Update weights
```

```python
plt.plot(range(epochs), losses)
plt.ylabel('Loss')
plt.xlabel('epoch');
plt.show()
```
